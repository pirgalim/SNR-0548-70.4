```python
import numpy as np
import matplotlib.pyplot as plt


oiii_curve = np.loadtxt("gmos_s_OIII_G0338_jun2017.txt", comments="#")
wavelength = oiii_curve[:,0]
transmission = oiii_curve[:,1]

oiiic_curve = np.loadtxt("gmos_s_OIIIC_G0339_jun2017.txt", comments="#")
wavelength_c = oiiic_curve[:,0]
transmission_c = oiiic_curve[:,1]

plt.plot(wavelength, transmission, color="red")
plt.plot(wavelength_c, transmission_c, color="blue")
plt.xlim(480, 540)
plt.xlabel("Wavelength (nm)")
plt.ylabel("Transmission")
plt.title("Gemini Filter Curve")
plt.grid()
plt.show()

print("OIII max: ", max(transmission))
print("OIIIC max: ", max(transmission_c))
```
