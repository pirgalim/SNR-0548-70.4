```python
import pandas as pd

cols = [
    "depth", "Te", "Htot", "hden", "eden", "2H2_H",
    "HI", "HII", "HeI", "HeII", "HeIII",
    "CO_C", "C1", "C2", "C3", "C4",
    "O1", "O2", "O3", "O4", "O5", "O6",
    "H2O_O", "AV_point", "AV_extend", "Tau912"
]

df = pd.read_csv("script.ovr", sep=r"\s+", comment="#", names=cols)

df.head()
```

```python
import matplotlib.pyplot as plt

plt.plot(df["depth"], df["Te"])
plt.xlabel("Depth (cm)")
plt.ylabel("Temperature (K)")
plt.show()
```

```python
plt.plot(df["depth"], df["O1"], label="O I")
plt.plot(df["depth"], df["O2"], label="O II")
plt.plot(df["depth"], df["O3"], label="O III")
plt.xlabel("Depth (cm)")
plt.ylabel("Ion fraction")
plt.legend()
plt.show()
```
